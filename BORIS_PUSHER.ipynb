```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def boris_push(r,v,q,m,dt,E_fuc,B_func):
    r_half=r+v*dt/2
    E=E_fuc(r_half)
    B=B_func(r_half)
    v_min=v+q*dt*E/(2*m)
    t_tor=q*dt*B/(2*m)
    v_prime = v_min + np.cross(v_min, t_tor)
    s_tor=2*t_tor/(1+t_tor*2)
    v_plu = v_min + np.cross(v_prime, s_tor)
    v_new=v_plu+q*dt*E/(2*m)
    r_new=r_half + v_new * dt/2
    return v_new,r_new
```

```python
def E_func(r):
    x, y, z = r
    return np.array([x**2, y, 0.0])
def B_func(r):
    x, y, z = r
    return np.array([0.0, 0.0, z])
```

```python
q=1
m=1
dt=0.05
r=np.array([1.0,0.0,0.0])
v=np.array([0.0,1.0,0.0])

N=1000
r_list=[]

for i in range(N):
    # if i==0:
        
    #     r_list.append(rv)
    #     v_list.append(vv)
    # else:
    #     rv,vv = boris_push(rv,vv,q,m,dt,E_func,B_func)
    #     r_list.append(rv)
    #     v_list.append(vv)
    r_list.append(r.copy())
    r,v= boris_push(r,v,q,m,dt,E_func,B_func)

r_list = np.array(r_list)
```

```python
plt.plot(r_list[:,0], r_list[:,1])
plt.xlabel("x")
plt.ylabel("y")
plt.gca().set_aspect("equal")

plt.show()
```
